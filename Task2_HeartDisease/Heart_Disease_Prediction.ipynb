## Introduction

##### **Definition** A binary classification ML task that predicts whether a patient has heart disease (Yes) or not (No) from four attributes collected during a cardiac check-up.
##### **Application** Supports early screening by flagging high-risk patients for further tests, and teaches how clinical test results relate to heart disease.

## Input and Output

### Inputs (Features / Attributes)

| **Input Feature/Attribute** | **UCI Name** | **Meaning** | **Possible Values** |
| --- | --- | --- | --- |
| **ChestPainType** | cp | Type of chest pain reported by the patient | Typical Angina, Atypical Angina, Non-Anginal Pain, Asymptomatic |
| **ExerciseInducedAngina** | exang | Chest pain during an exercise (treadmill) test | Yes, No |
| **MajorVessels** | ca | Number of major blood vessels coloured by fluoroscopy | Zero, One, Two, Three |

| **Output Attribute** | **Possible Values** |
| --- | --- |
| **HeartDisease** | Yes, No |

The UCI target `num` ranges from 0 to 4. Value 0 means no heart disease (**No**) and values 1 - 4 mean heart disease is present (**Yes**), the standard way this dataset is used.

### Step 2: Load Sample Data

The Titanic system started from a ready-made `sample-data.csv`. For heart disease, we build that file ourselves from the raw UCI data:
1. **Step 2.1** collects the raw file `processed.cleveland.data` (14 numeric columns, no header row).

<a id="step-2-1"></a>

#### Step 2.1: Data Collection - Get the Raw UCI File

````python
"""
Purpose of this section:
-------------------------
Collects the raw Cleveland heart disease data from the UCI Machine Learning
Repository. If 'processed.cleveland.data' is not already in this folder,
the official UCI zip file is downloaded and only that file is extracted.
"""

RAW_FILE = "processed.cleveland.data"
UCI_ZIP_URL = "https://archive.ics.uci.edu/static/public/45/heart+disease.zip"

if not os.path.exists(RAW_FILE):
    print("Downloading the dataset from UCI ...")
    urllib.request.urlretrieve(UCI_ZIP_URL, "heart+disease.zip")
    with zipfile.ZipFile("heart+disease.zip") as zf:
        zf.extract(RAW_FILE)
    os.remove("heart+disease.zip")

# The file has no header row, so we give the 14 columns their UCI names
uci_columns = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
               "thalach", "exang", "oldpeak", "slope", "ca", "thal", "num"]

# '?' marks a missing value in this file
raw_data = pd.read_csv(RAW_FILE, header=None, names=uci_columns, na_values="?")

print("\n\nRaw UCI Data (first 10 rows):")
print("=============================\n")
print(raw_data.head(10))
print("\nShape of raw data (rows, columns):", raw_data.shape)
````

<a id="step-2-2"></a>

#### Step 2.2: Create the Sample Data (4 Input Attributes + Output)

````python
"""
Purpose of this section:
-------------------------
Builds 'sample-data.csv' in the same style as the Titanic sample data:
1. Keep only 4 input attributes: cp, exang, ca, thal (and the output num).
2. Replace the numeric codes with text labels (codebook from the UCI page).
3. Convert the output num (0-4) into Yes / No.
Missing values stay empty here; they are handled in Step 3.2.
"""

chest_pain_codes = {1: "Typical Angina", 2: "Atypical Angina",
                    3: "Non-Anginal Pain", 4: "Asymptomatic"}
exercise_angina_codes = {1: "Yes", 0: "No"}
major_vessels_codes = {0: "Zero", 1: "One", 2: "Two", 3: "Three"}
thalassemia_codes = {3: "Normal", 6: "Fixed Defect", 7: "Reversible Defect"}

sample_data = pd.DataFrame({
    "ChestPainType": raw_data["cp"].map(chest_pain_codes),
    "ExerciseInducedAngina": raw_data["exang"].map(exercise_angina_codes),
    "MajorVessels": raw_data["ca"].map(major_vessels_codes),
    "Thalassemia": raw_data["thal"].map(thalassemia_codes),
    # num = 0 -> No heart disease, num = 1, 2, 3 or 4 -> heart disease present
    "HeartDisease": np.where(raw_data["num"] > 0, "Yes", "No"),
})

sample_data.to_csv("sample-data.csv", index=False, header=True)
print("Saved: sample-data.csv  ->", sample_data.shape[0], "rows,", sample_data.shape[1], "columns")
````

````python
"""
Purpose of this section:
-------------------------
Helps us understand the dataset by printing:
1. The names of all attributes (columns).
2. The total number of instances (rows).
3. The number of missing values in each attribute.
4. How many patients fall in each class and in each category.
"""

print("\n\nAttributes in Sample Data:")
print("==========================\n")
print(sample_data.columns)

print("\n\nNumber of Instances in Sample Data:", len(sample_data))
print("========================================\n")

print("Missing Values in Each Attribute:")
print(sample_data.isnull().sum())

print("\nClass Distribution of the Output (HeartDisease):")
print(sample_data["HeartDisease"].value_counts())

for attribute in ["ChestPainType", "ExerciseInducedAngina", "MajorVessels", "Thalassemia"]:
    print(f"\nValues of {attribute}:")
    print(sample_data[attribute].value_counts())
````

The chart below shows, for every value of each input attribute, the percentage of patients who have heart disease. A large gap between the bars of one attribute means that attribute is useful for the prediction.

<a id="step-3-2"></a>

#### Step 3.2: Pre-process Sample Data

- Step 3.1 shows missing values in **MajorVessels** (4 patients) and **Thalassemia** (2 patients). These came from the `?` entries in the raw UCI file.
- A Label Encoder cannot encode an empty value, and only 6 of 303 patients (about 2%) are affected, so these rows are **removed**.
- No other pre-processing is needed: all remaining values are valid text categories.

````python
"""
Purpose of this section:
-------------------------
Removes the instances (rows) that contain a missing value, then resets the
row numbers so they run from 0 again.
"""

rows_before = len(sample_data)

# Drop every row that has at least one missing value
sample_data = sample_data.dropna().reset_index(drop=True)

print("Instances before removing missing values:", rows_before)
print("Instances after removing missing values :", len(sample_data))
print("Rows removed:", rows_before - len(sample_data))
print("\nMissing values left:", int(sample_data.isnull().sum().sum()))
````

<a id="step-5"></a>

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

<a id="step-5-1"></a>

#### Step 5.1: Train the Label Encoder

````python
"""
Purpose of this section:
-------------------------
Sets up and trains one Label Encoder per attribute. A Label Encoder turns
text values (like 'Asymptomatic' or 'Reversible Defect') into numbers.

Steps:
1. List all possible values of each attribute.
2. Create a LabelEncoder for each attribute.
3. Train (fit) each LabelEncoder on its list of values.
"""

# All possible values of each attribute
chest_pain_type = pd.DataFrame({"ChestPainType": ["Typical Angina", "Atypical Angina",
                                                  "Non-Anginal Pain", "Asymptomatic"]})
exercise_induced_angina = pd.DataFrame({"ExerciseInducedAngina": ["Yes", "No"]})
major_vessels = pd.DataFrame({"MajorVessels": ["Zero", "One", "Two", "Three"]})
thalassemia = pd.DataFrame({"Thalassemia": ["Normal", "Fixed Defect", "Reversible Defect"]})
heart_disease = pd.DataFrame({"HeartDisease": ["Yes", "No"]})

# One LabelEncoder object per attribute
chest_pain_type_label_encoder = LabelEncoder()
exercise_induced_angina_label_encoder = LabelEncoder()
major_vessels_label_encoder = LabelEncoder()
thalassemia_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) the encoders; np.ravel() flattens each column into a 1D array
chest_pain_type_label_encoder.fit(np.ravel(chest_pain_type))
exercise_induced_angina_label_encoder.fit(np.ravel(exercise_induced_angina))
major_vessels_label_encoder.fit(np.ravel(major_vessels))
thalassemia_label_encoder.fit(np.ravel(thalassemia))
heart_disease_label_encoder.fit(np.ravel(heart_disease))

# Show the code given to each value (LabelEncoder numbers values in alphabetical order)
for name, encoder in [("ChestPainType", chest_pain_type_label_encoder),
                      ("ExerciseInducedAngina", exercise_induced_angina_label_encoder),
                      ("MajorVessels", major_vessels_label_encoder),
                      ("Thalassemia", thalassemia_label_encoder),
                      ("HeartDisease", heart_disease_label_encoder)]:
    print(f"{name:22s}", dict(zip(encoder.classes_, encoder.transform(encoder.classes_).tolist())))
````

<a id="step-5-2"></a>

#### Step 5.2: Label Encode the Output

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to the output attribute 'HeartDisease', so
'Yes' / 'No' become 1 / 0.
"""

# Copies of the dataset (one gets the encoded output, one keeps the original)
sample_data_encoded_output = sample_data.copy()
original_sample_data = sample_data.copy()

print("\n\nHeartDisease Attribute After Label Encoding:")
print("============================================\n")
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])

# Original text next to its numeric code
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

# Put the encoded output into the output-encoded copy
sample_data_encoded_output[["ChestPainType", "ExerciseInducedAngina", "MajorVessels", "Thalassemia", "HeartDisease"]] = \
    sample_data[["ChestPainType", "ExerciseInducedAngina", "MajorVessels", "Thalassemia", "encoded_heart_disease"]]

sample_data_encoded_output.to_csv("sample-data-encoded-output.csv", index=False, header=True)
print("\nSaved: sample-data-encoded-output.csv")
````

<a id="step-5-3"></a>

#### Step 5.3: Label Encode the Input

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to the 4 input attributes, so every column of the
dataset becomes numeric and can be given to the machine learning model.
"""

# Copy that will hold the fully encoded dataset
sample_data_encoded = sample_data_encoded_output.copy()

# Encode each input attribute with its own trained encoder
sample_data_encoded_output["encoded_chest_pain_type"] = chest_pain_type_label_encoder.transform(sample_data_encoded_output["ChestPainType"])
sample_data_encoded_output["encoded_exercise_induced_angina"] = exercise_induced_angina_label_encoder.transform(sample_data_encoded_output["ExerciseInducedAngina"])
sample_data_encoded_output["encoded_major_vessels"] = major_vessels_label_encoder.transform(sample_data_encoded_output["MajorVessels"])
sample_data_encoded_output["encoded_thalassemia"] = thalassemia_label_encoder.transform(sample_data_encoded_output["Thalassemia"])

# Replace the text columns with the encoded columns (the output is already encoded)
sample_data_encoded[["ChestPainType", "ExerciseInducedAngina", "MajorVessels", "Thalassemia", "HeartDisease"]] = \
    sample_data_encoded_output[["encoded_chest_pain_type", "encoded_exercise_induced_angina",
                                "encoded_major_vessels", "encoded_thalassemia", "HeartDisease"]]

print("\n\nSample Data after Label Encoding (first 20 rows):")
print("=================================================\n")
print(sample_data_encoded.head(20))

sample_data_encoded.to_csv("sample-data-encoded.csv", index=False, header=True)
print("\nSaved: sample-data-encoded.csv")
````

<a id="step-6"></a>

### Step 6: Execute the Training Phase

<a id="step-6-1"></a>

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data

80% Training Data and 20% Testing Data, as in the Titanic notebook. One change: the Titanic code uses `shuffle=False`, but here `shuffle=True` and `stratify` are used, because the UCI rows are not stored in random order. Stratifying keeps the same Yes/No ratio in both sets. `random_state=0` makes the split the same on every run.

````python
"""
Purpose of this section:
-------------------------
Splits the dataset into Training Data and Testing Data.
- Training Data teaches the model.
- Testing Data checks the model on examples it has never seen.
"""

training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,                                   # 20% for testing
    random_state=0,                                  # same split every run
    shuffle=True,                                    # mix rows before splitting
    stratify=sample_data_encoded["HeartDisease"]     # keep the Yes/No ratio in both sets
)

training_data_encoded.to_csv("training-data-encoded.csv", index=False, header=True)
testing_data_encoded.to_csv("testing-data-encoded.csv", index=False, header=True)

print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
print("\nHeart disease share in Training Data: {:.0%}".format(training_data_encoded["HeartDisease"].mean()))
print("Heart disease share in Testing Data : {:.0%}".format(testing_data_encoded["HeartDisease"].mean()))
````

<a id="step-6-2"></a>

### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

````python
"""
Purpose of this section:
-------------------------
Separates the input vectors (features) and the output labels (target)
of the Training Data.
"""

# First 4 columns = input attributes
input_vector_train = training_data_encoded.iloc[:, :-1]

# Last column = output label (HeartDisease)
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

<a id="step-6-3"></a>

### Step 6.3: Train the Support Vector Classifier

````python
"""
Purpose of this section:
-------------------------
Trains a Support Vector Classifier (SVC) on the Training Data. The SVC
learns the boundary that best separates patients with heart disease
from patients without it.
"""

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Same settings as the Titanic system
svc_model = svm.SVC(gamma="auto", random_state=0)

# Learn from the training inputs and their labels
svc_model.fit(input_vector_train, np.ravel(output_label_train))

print(svc_model)
````

<a id="step-6-4"></a>

### Step 6.4: Save the Trained Model

````python
"""
Purpose of this section:
-------------------------
Saves the trained SVC to a .pkl file, so the Testing Phase, the
Application Phase and the web app can use it without retraining.
"""

with open("heart_disease_svc_trained_model.pkl", "wb") as model_file:
    pickle.dump(svc_model, model_file)

print("Saved: heart_disease_svc_trained_model.pkl")
````

<a id="step-7"></a>

### Step 7: Execute the Testing Phase

<a id="step-7-1"></a>

#### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

````python
"""
Purpose of this section:
-------------------------
Separates the input vectors and the output labels of the Testing Data.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

<a id="step-7-2"></a>

### Step 7.2: Load the Saved Model

````python
"""
Purpose of this section:
-------------------------
Loads the saved SVC model from the .pkl file into memory.
"""

with open("heart_disease_svc_trained_model.pkl", "rb") as model_file:
    model = pickle.load(model_file)

print("Model loaded from heart_disease_svc_trained_model.pkl")
````

<a id="step-7-3"></a>

### Step 7.3: Evaluate the Machine Learning Model

<a id="step-7-3-1"></a>

#### Step 7.3.1: Make Predictions with the Trained Model on Testing Data

````python
"""
Purpose of this section:
-------------------------
Uses the trained model to predict the output for every patient in the
Testing Data and stores the predictions next to the true labels.
"""

model_predictions = model.predict(input_vector_test)

# Silence a harmless pandas warning about adding a column to a slice
pd.options.mode.chained_assignment = None
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv("model-predictions.csv", index=False, header=True)

model_predictions = testing_data_encoded
print("\n\nPredictions Returned by heart_disease_svc_trained_model (first 20 rows):")
print("========================================================================\n")
print(model_predictions.head(20))
````

<a id="step-7-4"></a>

### Step 7.4: Calculate the Accuracy Score

````python
"""
Purpose of this section:
-------------------------
Calculates the accuracy: the share of Testing Data patients the model
classified correctly.
"""

model_accuracy_score = accuracy_score(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
````

<a id="step-7-5"></a>

### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

````python
"""
Purpose of this section:
-------------------------
Accuracy alone hides which mistakes the model makes. For a medical task,
missing a sick patient matters more than a false alarm, so we also report:
- Precision: of the patients predicted 'Yes', how many really have heart disease?
- Recall: of the patients who really have heart disease, how many did the model find?
- F1-score: harmonic mean of precision and recall.
- Confusion Matrix: correct and wrong predictions for each class.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease (1)"]
))

conf_matrix = confusion_matrix(model_predictions["HeartDisease"], model_predictions["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No", "Actual: Yes"],
    columns=["Predicted: No", "Predicted: Yes"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", cbar=False,
            xticklabels=["No", "Yes"], yticklabels=["No", "Yes"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - SVC on Heart Disease Testing Data")
plt.tight_layout()
plt.show()
````
